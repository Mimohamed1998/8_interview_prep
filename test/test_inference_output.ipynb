```python
import sys
from pathlib import Path

PROJECT_ROOT = Path().resolve().parent
sys.path.insert(0, str(PROJECT_ROOT))

import pandas as pd
from src.common.utils.config import load_config

cfg = load_config()
OUTPUT_PATH = PROJECT_ROOT / cfg["inference"]["output_path"] / "predictions.parquet"

print(f"Path  : {OUTPUT_PATH}")
print(f"Exists: {OUTPUT_PATH.exists()}")
```

```python
df = pd.read_parquet(OUTPUT_PATH)

print(f"Shape      : {df.shape}")
print(f"Columns    : {df.columns.tolist()}")
print(f"year_month : {sorted(df['year_month'].unique())}")
print(f"cell_name  : {df['cell_name'].unique().tolist()}")
```

```python
print("=== Sample rows ===")
display(df.sample(10, random_state=42))

print("\n=== predicted_qty stats ===")
display(df["predicted_qty"].describe())
```
